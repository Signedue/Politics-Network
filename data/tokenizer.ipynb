```python
import pandas as pd
import spacy
from collections import Counter
from itertools import combinations
import networkx as nx

df = pd.read_csv("tv2_election_scraper/output/tv2_articles.csv")  
```

```python
# 2. Load the spaCy model 
# Note: Use 'da_core_news_sm' for Danish TV2 articles, or 'en_core_web_sm' if they are in English.
# Run 'python -m spacy download da_core_news_sm' in your terminal first if needed.
nlp = spacy.load("da_core_news_sm") 

def preprocess_text(text):
    if pd.isna(text):
        return []
    doc = nlp(str(text))

    tokens = [
        token.lemma_.lower() 
        for token in doc 
        if not token.is_stop and not token.is_punct and not token.is_space and token.lemma and len(token.text) > 2
    ]
    return tokens

df['tokens'] = df['body'].astype(str).apply(preprocess_text)
#sample_text = "Dette er en lille testtekst til at se, hvordan preprocessing fungerer, da jeg gik"
#print("Original:", sample_text)
#print("Processed Tokens:", preprocess_text(sample_text))
```
